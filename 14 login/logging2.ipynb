```python
import logging

logger1 = logging.getLogger('Module1')
logger1.setLevel(logging.DEBUG)

logger2 = logging.getLogger('Module2')
logger2.setLevel(logging.WARNING)
```

```python

logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    force=True
)
```

```python
logger1.debug("This is a message for logging debug")
logger2.warning("This is a message for logging Warnning ")
logger2.error("This is a message for logging error ")
```
